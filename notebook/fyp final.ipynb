```python
# Ordered Time
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

def parse_mixed_time_to_hour(series):
    # Convert mixed time formats to hour (0-23)
    def _to_hour(x):
        if pd.isna(x):
            return np.nan
        s = str(x).strip()

        # Excel decimal time (fraction of a day)
        try:
            v = float(s)
            if 0 <= v < 1:
                return int(v * 24)  # hour only
        except:
            pass

        # Normal time string (HH:MM or HH:MM:SS)
        t = pd.to_datetime(s, errors="coerce")
        if pd.isna(t):
            return np.nan
        return t.hour

    return series.apply(_to_hour)

bins = [0, 6, 12, 18, 24]
labels = ["0-6", "6-12", "12-18", "18-24"]

# ---- Time Ordered (EDA)
df["ordered_hour_eda"] = parse_mixed_time_to_hour(df["Time_Orderd"])
df["ordered_hour_group_eda"] = pd.cut(df["ordered_hour_eda"], bins=bins, labels=labels, right=False)

counts = df["ordered_hour_group_eda"].value_counts().sort_index()

plt.figure(figsize=(6,4))
ax = counts.plot(kind="bar")
plt.title("Orders by 6-Hour Interval (Time_Ordered)")
plt.ylabel("Number of Orders")
plt.xlabel("Hour Group")

for i, v in enumerate(counts.values):
    ax.text(i, v + max(counts)*0.01, str(int(v)), ha='center', fontsize=9)

plt.tight_layout()
plt.show()

print("Missing ordered_hour_eda:", df["ordered_hour_eda"].isna().sum())
print("Counts:\n", counts)
```

```python
# Parse Time Columns (Excel decimals + normal HH:MM)
import pandas as pd
import numpy as np

def parse_mixed_time(x):
    """
    Returns a pandas Timestamp anchored to 1900-01-01 with the correct HH:MM:SS,
    or NaT if cannot parse.
    NOTE: The 1900-01-01 date is a temporary placeholder only.
          It gets replaced by the real order_date in the next cell.
    Handles:
      - Excel decimal times (fraction of day): e.g. 0.458333333
      - Normal time strings: '22:35', '22:35:00'
    """
    if pd.isna(x):
        return pd.NaT

    s = str(x).strip()
    if s == "" or s.lower() == "nan":
        return pd.NaT

    # Excel decimal (fraction of a day)
    try:
        val = float(s)
        if 0 <= val < 1:
            total_sec = int(round(val * 86400))
            h   = total_sec // 3600
            m   = (total_sec % 3600) // 60
            sec = total_sec % 60
            s   = f"{h:02d}:{m:02d}:{sec:02d}"
    except:
        pass

    # Parse as time using a fixed anchor date (placeholder only)
    return pd.to_datetime("1900-01-01 " + s, errors="coerce")

df["time_orderd"]       = df["time_orderd"].apply(parse_mixed_time)
df["time_order_picked"] = df["time_order_picked"].apply(parse_mixed_time)

print("time_orderd missing      :", df["time_orderd"].isna().sum())
print("time_order_picked missing:", df["time_order_picked"].isna().sum())
print("\nSample parsed times:")
display(df[["time_orderd", "time_order_picked"]].head(5).apply(lambda col: col.dt.strftime("%H:%M:%S")))
```

```python
# Combine Date + Time into Full Timestamps + Midnight Fix
df["ordered_dt"] = pd.to_datetime(
    df["order_date"].dt.strftime("%Y-%m-%d") + " " +
    df["time_orderd"].dt.strftime("%H:%M:%S"),
    errors="coerce"
)
df["picked_dt"] = pd.to_datetime(
    df["order_date"].dt.strftime("%Y-%m-%d") + " " +
    df["time_order_picked"].dt.strftime("%H:%M:%S"),
    errors="coerce"
)

print("Combined order_date + time into ordered_dt / picked_dt")
print("ordered_dt missing:", df["ordered_dt"].isna().sum())
print("picked_dt  missing:", df["picked_dt"].isna().sum())

# Midnight crossing fix (handles orders placed just before midnight
mask_midnight = (
    df["picked_dt"].notna() &
    df["ordered_dt"].notna() &
    (df["picked_dt"] < df["ordered_dt"])
)
print("Midnight-crossing rows fixed:", int(mask_midnight.sum()))
df.loc[mask_midnight, "picked_dt"] += pd.Timedelta(days=1)

# Show final timestamps 
display(df[["order_date", "ordered_dt", "picked_dt"]].head(5))
```

Feature Engineering

```python
# Feature Engineering: Prep Time + Sanity Checks
df["prep_time_min"] = (df["picked_dt"] - df["ordered_dt"]).dt.total_seconds() / 60

neg = (df["prep_time_min"] < 0).sum()
print("Negative prep_time_min rows (should be 0 after midnight fix):", int(neg))

# Clip unreasonable values (optional but safe)
df["prep_time_min"] = df["prep_time_min"].clip(lower=0, upper=180)

print("prep_time_min summary:")
print(df["prep_time_min"].describe())
```

```python
# Feature Engineering: Temporal + Peak Hour

import pandas as pd
import matplotlib.pyplot as plt

# Create temporal features
df["order_hour"] = df["ordered_dt"].dt.hour
df["day_of_week"] = df["order_date"].dt.dayofweek
df["is_weekend"] = (df["day_of_week"] >= 5).astype(int)

# Step 1: Compute hour counts for table and chart
hour_counts = df["order_hour"].value_counts().sort_index()

# Step 2: Display clean hourly order table
hour_table = pd.DataFrame({
    "Time Range": [f"{int(h):02d}:00 - {int(h):02d}:59" for h in hour_counts.index],
    "No. of Orders": hour_counts.values,
    "% of Total": (hour_counts.values / hour_counts.sum() * 100).round(1)
})

hour_table.index = range(len(hour_table))
display(hour_table)

# Step 3: Create peak_hour category
# Morning rush : 08:00–11:59
# Off-peak     : 12:00–16:59 and other non-peak hours
# Evening peak : 17:00–23:59

def time_category(hour):
    if pd.isna(hour):
        return pd.NA
    elif 8 <= hour <= 11:
        return "morning_rush"
    elif 17 <= hour <= 23:
        return "evening_peak"
    else:
        return "off_peak"

df["peak_hour"] = df["order_hour"].apply(time_category)

# Step 4: Visualise orders by hour with category colours
def get_color(hour):
    if 8 <= hour <= 11:
        return "#f39c12"      # orange = morning rush
    elif 17 <= hour <= 23:
        return "#e74c3c"      # red = evening peak
    else:
        return "steelblue"   # blue = off-peak

colors = [get_color(h) for h in hour_counts.index]

plt.figure(figsize=(10, 4))
plt.bar(hour_counts.index, hour_counts.values, color=colors)
plt.title("Orders by Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Order Count")
plt.xticks(hour_counts.index, rotation=0)
plt.tight_layout()
plt.show()

# Step 5: Print confirmation and category distribution
print("Temporal features created successfully.")

print("\nPeak hour category distribution:")
print(df["peak_hour"].value_counts().sort_index())
```

```python
# Feature Engineering: Distance (Haversine) + Checks
from numpy import radians, sin, cos, sqrt, arctan2

# Convert to radians
lat1 = radians(df["restaurant_latitude"])
lon1 = radians(df["restaurant_longitude"])
lat2 = radians(df["delivery_location_latitude"])
lon2 = radians(df["delivery_location_longitude"])

dlat = lat2 - lat1
dlon = lon2 - lon1

a = sin(dlat / 2) ** 2 + cos(lat1) * cos(lat2) * sin(dlon / 2) ** 2
c = 2 * arctan2(sqrt(a), sqrt(1 - a))

R = 6371  # Earth radius in km
df["distance_km"] = R * c

df["distance_km"] = df["distance_km"].clip(lower=0, upper=50)

print("Distance feature created (vectorized)")
print(df["distance_km"].describe())
```

# Phase 6: Feature Selection

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import sparse
from sklearn.feature_selection import SelectKBest, mutual_info_regression, RFE
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import cross_val_score
from xgboost import XGBRegressor

# Get feature names after preprocessing

feature_names = preprocessor_tree.get_feature_names_out()
feature_names = np.array(feature_names)

print("Total features after preprocessing:", len(feature_names))

# Convert sparse matrix to dense array if needed
if sparse.issparse(X_train_tree_processed):
    X_train_fs = X_train_tree_processed.toarray()
else:
    X_train_fs = X_train_tree_processed

if sparse.issparse(X_test_tree_processed):
    X_test_fs = X_test_tree_processed.toarray()
else:
    X_test_fs = X_test_tree_processed

print("X_train_fs shape:", X_train_fs.shape)
print("X_test_fs shape :", X_test_fs.shape)


# Safety check: missing values after imputation
print("Missing values after preprocessing:")
print("X_train_fs:", np.isnan(X_train_fs).sum())
print("X_test_fs :", np.isnan(X_test_fs).sum())


# Mutual Information Feature Ranking
selector_mi_all = SelectKBest(score_func=mutual_info_regression, k="all")
selector_mi_all.fit(X_train_fs, y_train)

mi_scores = selector_mi_all.scores_
sorted_idx = np.argsort(mi_scores)[::-1]

feature_scores = pd.DataFrame({
    "Rank": range(1, len(feature_names) + 1),
    "Feature": feature_names[sorted_idx],
    "MI Score": mi_scores[sorted_idx]
})

feature_scores["MI Score"] = feature_scores["MI Score"].round(4)

print("Mutual Information scores computed on preprocessed training data only.")
print(f"Total features scored: {len(feature_scores)}\n")

strong = feature_scores[feature_scores["MI Score"] >= 0.04]
weak = feature_scores[feature_scores["MI Score"] < 0.04]

print(f"── Strong features (MI ≥ 0.04): {len(strong)} ──")
display(strong.reset_index(drop=True))

print(f"\n── Weak features (MI < 0.04): {len(weak)} ──")
display(weak.reset_index(drop=True))


# Plot Top 15 MI features
top15 = feature_scores.head(15)

colors = [
    "#2ecc71" if score >= 0.04 else "steelblue"
    for score in top15["MI Score"]
]

plt.figure(figsize=(10, 6))
plt.barh(top15["Feature"], top15["MI Score"], color=colors)
plt.xlabel("Mutual Information Score")
plt.title("Top 15 Features by Mutual Information\nGreen = Strong ≥ 0.04, Blue = Weak")
plt.gca().invert_yaxis()
plt.axvline(x=0.04, color="red", linestyle="--", linewidth=1, label="Threshold = 0.04")
plt.legend()
plt.tight_layout()
plt.show()

```

RF Score (Random Forest Feature Importance)
- What it measures: How much a feature reduces prediction error when used as a split point across all trees in the Random Forest.
- How it works: When RF builds decision trees, it tracks how much each feature improves the split at each node. Features used more often and at higher nodes get a higher score.
- Scale: 0 to 1, and all scores add up to exactly 1.0
- RF Score = 0.209  → feature is responsible for 20.9% of prediction improvement in the forest


MI Score (Mutual Information)
- What it measures: How much information a feature shares with the target variable (time_taken_min). Completely independent of any model.
- How it works: Measures statistical dependency — if knowing the feature value helps you predict the delivery time, MI score is high. If knowing it tells you nothing, MI score is 0.
- Scale: 0 and above (no upper limit, but typically 0–0.2 in practice)
- MI Score = 0.133  → knowing order_hour gives you significant information about delivery time

RF Feature Importance (Embedded Method) was selected as the final feature selection method. To address the known limitation of RF importance underscoring correlated features, a hybrid threshold criterion was applied — retaining features with either RF Score ≥ 0.015 or MI Score ≥ 0.04. This ensured that features such as order_hour (MI=0.133) and road_traffic_density_Jam (MI=0.067) were correctly retained despite receiving lower RF importance scores due to feature correlation

```python
# LightGBM Optuna CV
import optuna
from lightgbm import LGBMRegressor
from sklearn.model_selection import cross_val_score

optuna.logging.set_verbosity(optuna.logging.WARNING)

def objective_lgbm_cv(trial):
    params = {
        'n_estimators'    : trial.suggest_int('n_estimators', 100, 700),
        'learning_rate'   : trial.suggest_float('learning_rate', 0.01, 0.3),
        'max_depth'       : trial.suggest_int('max_depth', 3, 10),
        'num_leaves'      : trial.suggest_int('num_leaves', 20, 100),
        'subsample'       : trial.suggest_float('subsample', 0.6, 1.0),
        'colsample_bytree': trial.suggest_float('colsample_bytree', 0.6, 1.0),
        'reg_alpha'       : trial.suggest_float('reg_alpha', 0.0, 1.0),
        'reg_lambda'      : trial.suggest_float('reg_lambda', 0.5, 5.0),
        'min_child_samples': trial.suggest_int('min_child_samples', 5, 50)
    }

    model = LGBMRegressor(
        **params,
        random_state=42,
        verbose=-1
    )

    scores = cross_val_score(
        model,
        X_train_hybrid,
        y_train,
        cv=5,
        scoring='neg_root_mean_squared_error',
        n_jobs=-1
    )

    return scores.mean()  # negative RMSE — Optuna maximises

study_lgbm_cv = optuna.create_study(
    direction='maximize',
    sampler=optuna.samplers.TPESampler(seed=42)
)

study_lgbm_cv.optimize(objective_lgbm_cv, n_trials=100, show_progress_bar=True)

print("\nBest CV RMSE :", round(-study_lgbm_cv.best_value, 4))
print("Best Params  :", study_lgbm_cv.best_params)

tuned_lgbm_optuna_cv = LGBMRegressor(
    **study_lgbm_cv.best_params,
    random_state=42,
    verbose=-1
)

tuned_lgbm_optuna_cv.fit(X_train_hybrid, y_train)
print("\ntuned_lgbm_optuna_cv saved successfully.")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
import pandas as pd

def evaluate_model(model, X_train, X_test, y_train, y_test, model_name):
    y_pred_train = model.predict(X_train)
    y_pred_test  = model.predict(X_test)

    return {
        "Model"      : model_name,
        "Train MAE"  : round(mean_absolute_error(y_train, y_pred_train), 4),
        "Test MAE"   : round(mean_absolute_error(y_test,  y_pred_test),  4),
        "Train RMSE" : round(np.sqrt(mean_squared_error(y_train, y_pred_train)), 4),
        "Test RMSE"  : round(np.sqrt(mean_squared_error(y_test,  y_pred_test)),  4),
        "Train R²"   : round(r2_score(y_train, y_pred_train), 4),
        "Test R²"    : round(r2_score(y_test,  y_pred_test),  4),
    }

results = []

# --- Baseline models ---
results.append(evaluate_model(lr,           X_train_lr_t,           X_test_lr_t,           y_train, y_test, "Linear Regression (Baseline)"))
results.append(evaluate_model(dt,           X_train_tree_processed, X_test_tree_processed, y_train, y_test, "Decision Tree (Baseline)"))
results.append(evaluate_model(rf,           X_train_tree_processed, X_test_tree_processed, y_train, y_test, "Random Forest (Baseline)"))
results.append(evaluate_model(baseline_xgb, X_train_tree_processed, X_test_tree_processed, y_train, y_test, "XGBoost (Baseline)"))  
results.append(evaluate_model(baseline_lgbm,X_train_tree_processed, X_test_tree_processed, y_train, y_test, "LightGBM (Baseline)"))


# --- Tuned: Linear Regression ---
results.append(evaluate_model(tuned_lr,               X_train_lr_t, X_test_lr_t, y_train, y_test, "Linear Regression (Tuned - Ridge GridSearchCV)"))

# --- Tuned: Decision Tree ---
results.append(evaluate_model(tuned_dt,               X_train_hybrid,    X_test_hybrid,    y_train, y_test, "Decision Tree (Tuned - GridSearchCV)"))

# --- Tuned: Random Forest ---
results.append(evaluate_model(tuned_rf,               X_train_hybrid,    X_test_hybrid,    y_train, y_test, "Random Forest (Tuned - RandomizedSearchCV)"))
results.append(evaluate_model(tuned_rf_optuna,        X_train_hybrid,    X_test_hybrid,    y_train, y_test, "Random Forest (Tuned - Optuna)"))

# --- Tuned: XGBoost ---
results.append(evaluate_model(tuned_xgb_rand,         X_train_hybrid,    X_test_hybrid,    y_train, y_test, "XGBoost (Tuned - RandomizedSearchCV)"))
results.append(evaluate_model(tuned_xgb_optuna_cv,    X_train_hybrid,    X_test_hybrid,    y_train, y_test, "XGBoost (Tuned - Optuna CV)"))
results.append(evaluate_model(tuned_xgb_optuna_prune, X_train_hybrid,    X_test_hybrid,    y_train, y_test, "XGBoost (Tuned - Optuna + Pruning)"))

# --- Tuned: LightGBM ---
results.append(evaluate_model(tuned_lgbm_rand,      X_train_hybrid, X_test_hybrid, y_train, y_test, "LightGBM (Tuned - RandomizedSearchCV)"))
results.append(evaluate_model(tuned_lgbm_optuna_cv, X_train_hybrid, X_test_hybrid, y_train, y_test, "LightGBM (Tuned - Optuna CV)"))

results_df = pd.DataFrame(results)

# Sort by Test RMSE ascending — best model at top
results_df = results_df.sort_values("Test RMSE", ascending=True).reset_index(drop=True)
results_df.index += 1

print("=== Model Comparison: Baseline vs Tuned (Sorted by Test RMSE) ===")
display(results_df)
```

```python
# Overfitting check
import pandas as pd

print("=== Overfitting Check (Train vs Test) ===\n")

# Calculate thresholds dynamically from your target variable
r2_threshold   = 0.05
rmse_threshold = round(y.mean() * 0.10, 4)

print(f"R² Gap threshold  : > {r2_threshold}       (fixed scale 0–1)")
print(f"RMSE Gap threshold: > {rmse_threshold} min  (10% of target mean {round(y.mean(), 2)} min)")
print()

# Build overfitting dataframe
overfit_df = results_df[[
    "Model",
    "Train RMSE", "Test RMSE",
    "Train R²",   "Test R²"
]].copy()

# Calculate gaps
overfit_df["RMSE Gap (Test-Train)"] = (overfit_df["Test RMSE"] - overfit_df["Train RMSE"]).round(4)
overfit_df["R² Gap (Train-Test)"]   = (overfit_df["Train R²"]  - overfit_df["Test R²"]).round(4)

# R² status
overfit_df["R² Status"] = overfit_df["R² Gap (Train-Test)"].apply(
    lambda x: "Overfit"  if x >  r2_threshold  else
              "Underfit" if x < -r2_threshold  else
              "Good fit"
)

# RMSE status
overfit_df["RMSE Status"] = overfit_df["RMSE Gap (Test-Train)"].apply(
    lambda x: "Overfit"  if x >  rmse_threshold  else
              "Underfit" if x < -rmse_threshold  else
              "Good fit"
)

# Combined verdict — overfit only if BOTH agree
def combined_verdict(row):
    if row["R² Status"] == "Overfit"  and row["RMSE Status"] == "Overfit":
        return "Overfit"
    elif row["R² Status"] == "Underfit" and row["RMSE Status"] == "Underfit":
        return "Underfit"
    elif row["R² Status"] == "Overfit"  and row["RMSE Status"] == "Good fit":
        return "Borderline"
    else:
        return "Good fit"

overfit_df["Final Verdict"] = overfit_df.apply(combined_verdict, axis=1)

# Sort by R² gap descending
overfit_df = overfit_df.sort_values("R² Gap (Train-Test)", ascending=False).reset_index(drop=True)
overfit_df.index += 1

# Color mapping
def color_status(v):
    if v == "Overfit"    : return "color: #c0392b; font-weight:bold"
    if v == "Borderline" : return "color: #e67e22; font-weight:bold"
    if v == "Underfit"   : return "color: #8e44ad; font-weight:bold"
    if v == "Good fit"   : return "color: #27ae60; font-weight:bold"
    return ""

display(overfit_df.style
    .map(color_status, subset=["R² Status", "RMSE Status", "Final Verdict"])
    .format({
        "Train RMSE"          : "{:.4f}",
        "Test RMSE"           : "{:.4f}",
        "Train R²"            : "{:.4f}",
        "Test R²"             : "{:.4f}",
        "RMSE Gap (Test-Train)": "{:.4f}",
        "R² Gap (Train-Test)" : "{:.4f}"
    })
)

# Summary count
print("\nFinal Verdict Summary:")
print(overfit_df["Final Verdict"].value_counts().to_string())
```

```python
import numpy as np
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

X_test_df = X_test_tree.copy()
X_test_df['y_true'] = y_test.values
X_test_df['y_pred'] = tuned_lgbm_optuna_cv.predict(X_test_hybrid)

def evaluate_subset(name, mask):
    subset = X_test_df[mask]
    if len(subset) < 10:
        print(f"{name}: insufficient data ({len(subset)} rows)")
        return
    rmse = np.sqrt(mean_squared_error(subset['y_true'], subset['y_pred']))
    mae  = mean_absolute_error(subset['y_true'], subset['y_pred'])
    r2   = r2_score(subset['y_true'], subset['y_pred'])
    print(f"{name:35s} | n={len(subset):5d} | RMSE={rmse:.4f} | MAE={mae:.4f} | R²={r2:.4f}")

print("=" * 80)
print("ROBUSTNESS ANALYSIS — LightGBM (Tuned - Optuna CV)")
print("=" * 80)

print("\n── Peak Hour Subsets ──")
evaluate_subset("Morning Rush (8–11)",    X_test_df['order_hour'].between(8, 11))
evaluate_subset("Off-Peak (12–16)",       X_test_df['order_hour'].between(12, 16))
evaluate_subset("Evening Peak (17–23)",   X_test_df['order_hour'].between(17, 23))

print("\n── Weekend vs Weekday ──")
evaluate_subset("Weekday (Mon–Fri)",      X_test_df['is_weekend'] == 0)
evaluate_subset("Weekend (Sat–Sun)",      X_test_df['is_weekend'] == 1)

print("\n── Festival vs Non-Festival ──")
evaluate_subset("Festival Period",        X_test_df['festival'] == 'Yes')
evaluate_subset("Non-Festival Period",    X_test_df['festival'] == 'No')

print("\n── Traffic Density Subsets ──")
evaluate_subset("Low Traffic",            X_test_df['road_traffic_density'] == 'Low')
evaluate_subset("Medium Traffic",         X_test_df['road_traffic_density'] == 'Medium')
evaluate_subset("High Traffic",           X_test_df['road_traffic_density'] == 'High')
evaluate_subset("Jam Traffic",            X_test_df['road_traffic_density'] == 'Jam')

print("\n── Overall Test Set ──")
evaluate_subset("Full Test Set",          pd.Series([True] * len(X_test_df), index=X_test_df.index))
```

Evening peak hours (17–23) showed the strongest performance (R² = 0.8395), which aligns with the majority of training samples belonging to this period. Morning rush hours (8–11) showed relatively lower performance (R² = 0.7517), likely due to fewer training samples in that time window.
The most significant limitation was identified during festival periods, where R² dropped to 0.3538 despite a low RMSE of 3.31 minutes. This is attributed to only 179 festival records in the test set (2.1% of total), which is insufficient for reliable evaluation and reflects the class imbalance in festival data. For traffic conditions, the model performed well on jam and medium traffic but showed slightly weaker performance on high traffic (R² = 0.7471), suggesting variability in high traffic delivery patterns.
Overall, the robustness analysis confirms that the model generalises reliably across most operational conditions, with the exception of festival periods which represent a known limitation due to data imbalance.
